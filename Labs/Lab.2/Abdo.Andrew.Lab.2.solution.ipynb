*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def all_same(a_list):
    if len(a_list)==0:
        return False
    for item in a_list[1:]:
        if a_list[0]!=item:
            return False
    return a_list[0]

def transpose(M):
    n_rows = len(M)
    n_columns = len(M[0])

    out = [[0]*n_rows for _ in range(n_columns)]

    for i in range(len(M)):
        for j in range(len(M[i])):
            out[j][i]=M[i][j]

    return out

def get_diagonals(matrix):
    n=len(matrix)
    out_1= list()
    out_2= list()
    for i in range(n):
        out_1.append(matrix[i][i])
        out_2.append(matrix[i][n-i-1])
    return [out_1,out_2]

def check_empty(mat):
    missing = 0
    for i in range(len(mat)):
        for j in range(len(mat[i])):
            if (mat[i][j] == 0):
                missing = -1
    return missing
            

def game_state(board):
    columns= list(map(all_same,board))
    rows = list(map(all_same,transpose(board)))
    diagonals = list(map(all_same,get_diagonals(board)))
    for i in range(len(board)):
        if (columns[i] != False):
            return (columns[i])
        elif (rows[i] != False):
            return (rows[i])
        elif (diagonals[0] != False):
            return (diagonals[0])
        elif (diagonals[1] != False):
            return (diagonals[1])
    return check_empty(board)
            
                  
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Write your solution here
def game_play(board,player,location):
    x,y = location
    if (board[x][y] == 0):
        board[x][y] = player
        return True
    else:
        return False
    
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
# Write your solution here
def parse_location(l_string):
    if not isinstance(l_string,str):
        print("Bad Input. Location must be string.")
        return False
    
    if len(l_string)!=2:
        print("Bad Input. Location must be 2 characters.")
        return False
    
    row=l_string[0].upper()
    col=l_string[1].upper()
    
    if row not in row_map:
        print("Bad Row.")
        return False

    if col not in column_map:
        print("Bad Column.")
        return False

    return row_map[row],column_map[col]
    
def human_play(board,player,location):
    loc=parse_location(location)
    if loc !=-1:
        return game_play(board,player,loc)
    else:
        print("Bad move.")
        return False 
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
# Write yourrr solution here
def tic_tac_toe_game(size):
    print("welcome to tic_tac_toe.")
    print ("--------------------")

    game_board = blank_board(size)

    player = player_1

    game_won = False
    while not game_won:
        draw_game_labeled(game_board)

        print("player ",player," move:")
        get_move(game_board,player)

        if (game_state(game_board) != -1):
            game_won = True

        if player==player_1:
            player=player_2
        else:
            player=player_1
            
    draw_game_labeled(game_board)
    if (game_state(game_board) == 0):
        print ("the game ended in a tie")
    else:
        print("Winner is player:",game_state(game_board))
        

    
````
